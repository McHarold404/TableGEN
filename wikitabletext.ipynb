```python

def parse_table2(table_text: str) -> tuple[dict[str, str], pd.DataFrame]:
    """
    Parses a table string to extract title, subtitle, and key-value pairs.
    
    Args:
        table_text (str): The table as a string with <NEWLINE> as line separators.
    
    Returns:
        Tuple[Dict[str, str], pd.DataFrame]: 
            - A dictionary containing 'title' and/or 'subtitle' if they exist.
            - A pandas DataFrame with two columns: 'Key' and 'Value'.
    """
    # Initialize storage
    header: Dict[str, str] = {}
    data: list = []
    header_extracted = False  # Flag to indicate if header has been fully extracted
    
    # Replace <NEWLINE> with actual newline characters
    table_text = table_text.replace('<NEWLINE>', '\n')
    
    # Split the text into lines
    lines = table_text.split('\n')
    
    for line in lines:
        # Use regex to find all key-value pairs in the format | key | value |
        # This regex captures pairs like | key | value |
        matches = re.findall(r'\|\s*([^|]+?)\s*\|\s*([^|]+?)\s*(?=\||$)', line)
        
        for key, value in matches:
            key_clean = key.strip().lower()  # Normalize key to lowercase
            value_clean = value.strip()
            
            if not header_extracted and key_clean in ['title', 'subtitle']:
                # Extract header information
                header[key_clean] = value_clean
            else:
                # Add to data regardless of key name
                data.append({'Key': key_clean, 'Value': value_clean})
        
        # Determine if header extraction is complete
        # If any non-title/subtitle key is found, stop header extraction
        if not header_extracted:
            non_header_keys = [k.strip().lower() for k, v in matches if k.strip().lower() not in ['title', 'subtitle']]
            if non_header_keys:
                header_extracted = True
    
    # Create DataFrame from data
    df = pd.DataFrame(data)
    
    return header, df
```

```python
import pandas as pd
import re
from typing import List, Dict

def parse_wikibio(data: str) -> pd.DataFrame:
    """
    Parses a text string containing key-value pairs separated by '|'.
    
    Args:
        data (str): The raw input text containing key-value pairs separated by '|'.
        
    Returns:
        pd.DataFrame: A DataFrame with two columns: 'Key' and 'Value'.
    """
    # Split the input text into lines based on the literal string <NEWLINE>
    lines = data.split('<NEWLINE>')
    
    # Initialize a list to store key-value pairs
    data_list = []
    
    # Process each line to extract key-value pairs
    for line_number, line in enumerate(lines, 1):
        # Strip leading/trailing whitespace
        line = line.strip()
        if not line:
            continue  # Skip empty lines

        
        # Use regex to find all key-value pairs in the format | key | value |
        # This regex captures pairs like | key | value | or | key | value
        matches = re.findall(r'\|\s*([^|]+?)\s*\|\s*([^|]+?)\s*(?:\||$)', line)
        
        # Add each pair to the data list
        for key, value in matches:
            data_list.append({'Key': key.strip(), 'Value': value.strip()})
    
    # Create a DataFrame from the extracted key-value pairs
    df = pd.DataFrame(data_list, columns=['Key', 'Value'])
    
    return df
```
